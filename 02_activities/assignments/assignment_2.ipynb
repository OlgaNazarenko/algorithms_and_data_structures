```python
# Definition for a binary tree node.
from collections import Counter

# traverse left subtree
# check root value
# traverse right subtree

class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

def is_duplicate3(root: TreeNode) -> int:
    def traverse(node, seen, duplicates):
        if not node:
            return -1
        
        traverse(node.left, seen, duplicates)
        
        if node.val in seen:
            duplicates.append(node.val)
        else:
            seen.append(node.val)

        traverse(node.right, seen, duplicates)

    seen = []
    duplicates = []
    traverse(root, seen, duplicates)

    return list(duplicates)[0] if duplicates else -1
```

  ## Examples

  Output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]]

  ### Example 2

  ![](./images/q1_ex3.png)

  Input: `root = [10, 9, 7, 8]`

  Output: [[10, 7], [10, 9, 8]]

</details>

#### Starter Code for Question 2

```python
# Definition for a binary tree node.

from typing import List

class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right
        
def bt_path(root: TreeNode) -> List[List[int]]:
    def traverse(node, path, paths): 
        if not node: # a base case, to stop recursion
            return
        
        # to build all paths to each val
        path.append(node.val)
        
        # to check if the node is a leaf, has its children, and if both children are empty
        if node.left is None and node.right is None:
            # to add the copy of this path, as later we take one step back - to the parent node
            paths.append(list(path)) 
        else:
            # traverse both sides
            traverse(node.left, path, paths)
            traverse(node.right, path, paths)
        
        # to go one step back, to see other branches.
        path.pop()
    
    path = []
    paths = []
    traverse(root, path, paths)
    return paths    
```

```python
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)
```

```python
print(bt_path(root))
```

  # Question Three: Missing Number in Range
 
  You are given a list containing `n` integers in the range `[0, n]`. Return a list of numbers that are missing from the range `[0, n]` of the array. If there is no missing number, return -1. Note, all the integers in the list may not be unique.
  
  ## Examples

  ### Example 1

  Input: `lst = [0, 2]`

  Output: [1]

  ### Example 2

  Input: `lst = [5, 0, 1]`

  Output: [2, 3, 4]

  ### Example 3

  Input: `lst = [6, 8, 2, 3, 5, 7, 0, 1, 10]`

  Output: [4, 9]

</details>

#### Starter Code for Question 3


```python
def missing_num(nums: List) -> int:
    
    num_set = set(nums) # hash table

    # We generate a full list of expected numbers starting from 0 to the maximum value in nums
    lst = [a for a in range(min(nums), max(nums)+1)]
    
    missing_numbers = []
    
    for num in lst:
        if num not in num_set:
            missing_numbers.append(num)
        
    return missing_numbers or -1

# space and time complexity - o(n)
```

```python
lst1 = [0, 1, 2, 3] # should return -1

missing_num(lst1)
```

```python
lst2 = [5, 0, 1] # should return [2, 3, 4]

missing_num(lst2)
```

```python
lst4 = [5, 1] # should return [2, 3, 4]

missing_num(lst4)
```

```python
lst3 = [6, 8, 2, 3, 5, 7, 0, 1, 10] # should return [4, 9]

missing_num(lst3)
```


-   Copy the solution your partner wrote. 


```python
# Your answer here
from typing import List

class Solution:
    def move_zeros_to_end(self, nums: List[int]) -> None:
        """
        Moves all zeros to the end of the list in-place while
        maintaining the relative order of the non-zero elements.
        """
        left = 0
        for right in range(len(nums)):
            if nums[right] != 0:
                nums[left], nums[right] = nums[right], nums[left]
                left += 1

# Example usage:
solution = Solution()

# Example 1
nums1 = [0, 1, 0, 3, 12]
solution.move_zeros_to_end(nums1)
print(f"Input: [0, 1, 0, 3, 12], Output: {nums1}")
```
