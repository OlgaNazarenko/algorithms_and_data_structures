## Question Two: Valid Bracket Sequence
**Description**  
Given a string containing only the characters `'('`, `')'`, `'{'`, `'}'`, `'['`, and `']'`, determine if the input string is a **valid bracket sequence**.  
A string is valid if:
- Open brackets are closed by the same type of brackets, and
- Open brackets are closed in the correct order.

**Examples**
```python
Input: s = "([]{})"
Output: True
```
```python
Input: s = "([)]"
Output: False
```
```python
Input: s = "()[]{}"
Output: True
```
```python
Input: s = "[{]}"
Output: False
```

**Question 2 Starter Code**

````python
def is_valid_brackets(s: str) -> bool:
    # opening_brackets = "({["
    # closing_brackets = ")}]"

    stack = []

    for el in s:
        if el == "(":
            stack.append(")")
        elif el == "{":
            stack.append('}')
        elif el == "[":
            stack.append("]")
        elif not stack or el != stack.pop():
            return False
    return len(stack) == 0

        
    
````

````python
s1 = "([]{})"
print(is_valid_brackets(s1))

s2 = "([)]"
print(is_valid_brackets(s2))

s3 = "()[]{}"
print(is_valid_brackets(s3))

s4 = "[{]}"
print(is_valid_brackets(s4))
````

## Question Three: Move All Zeros to End
**Description**  
Given a list of integers, move all zeros to the end while maintaining the relative order of the non-zero elements. 

**Examples**
```python
Input: nums = [0, 1, 0, 3, 12]
Output: [1, 3, 12, 0, 0]
```
```python
Input: nums = [4, 0, 5, 0, 0, 6]
Output: [4, 5, 6, 0, 0, 0]
```


````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    gen_numbers = []
    zeros_numbers = []

    for char in nums:
        if char != 0:
            gen_numbers.append(char)
        else:
            zeros_numbers.append(0)
            # print(f'N of zeros: {zeros_numbers.count(0)}')
    # gen_numbers.extend(zeros_numbers)

    return gen_numbers + zeros_numbers
````

````python
nums5 = [0, 1, 0, 3, 12]
print(move_zeros_to_end(nums5))

nums6 = [4, 0, 5, 0, 0, 6]
print(move_zeros_to_end(nums6))
````

````python
## version 2.
from typing import List

def move_zeros_to_end3(nums: List[int]) -> List[int]:
    return [n for n in nums if n != 0] + [n for n in nums if n ==0 ] 
````

````python
# Your answer here

from typing import List
from collections import Counter

def first_duplicate(nums: List[int]) -> int:
    duplicates = []
    seen = set()

    for num in nums:
        if num in seen:
            duplicates.append(num)
            return num
        seen.add(num)
    return -1

    # return duplicates
````

````python
nums = [1, 2, 3, 4, 5, 7, 6]

first_duplicate(nums)
````

````python
nums2 = [3, 2, 3, 2, 5, 7, 6]

first_duplicate(nums2)
````

````python
import timeit

lst_time = timeit.timeit(lambda: first_duplicate(nums), number=1)
print(f"It took: {lst_time:.6f} seconds")
````
